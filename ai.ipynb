```python
# Prime number check
num = int(input("Enter a number: "))

if num > 1:
    for i in range(2, num):
        if num % i == 0:
            print(num, "is not a prime number")
            break
    else:
        print(num, "is a prime number")
else:
    print(num, "is not a prime number")
```

```python
list1 = list(map(int, input("Enter numbers for list1 (space-separated): ").split()))
list2 = list(map(int, input("Enter numbers for list2 (space-separated): ").split()))

print("Length of list1:", len(list1))

list3 = list1 + list2
print("Concatenated list:", list3)

num1 = int(input("Enter a number to check in list1: "))
print(f"Is {num1} in list1?", num1 in list1)

num2 = int(input("Enter another number to check in list1: "))
print(f"Is {num2} in list1?", num2 in list1)

print("Iteration of list2:")
for item in list2:
    print(item)

index = int(input("Enter index to access list1: "))
if 0 <= index < len(list1):
    print(f"Indexing list1[{index}]:", list1[index])
else:
    print("Invalid index! Out of range.")

start = int(input("Enter slice start index for list3: "))
end = int(input("Enter slice end index for list3: "))
if 0 <= start < len(list3) and 0 <= end <= len(list3):
    print(f"Slicing list3[{start}:{end}]:", list3[start:end])
else:
    print("Invalid slice range! Out of range.")
```

```python
lst = list(map(int, input("Enter numbers for list (space-separated): ").split()))

lst.append(int(input("Enter a number to append: ")))
print("After append:", lst)

lst.extend(list(map(int, input("Enter numbers to extend (space-separated): ").split())))
print("After extend:", lst)

pos = int(input("Enter index to insert: "))
val = int(input("Enter value to insert: "))
lst.insert(pos, val)
print("After insert:", lst)

rem = int(input("Enter value to remove: "))
if rem in lst:
    lst.remove(rem)
    print("After remove:", lst)
else:
    print("Value not found!")

idx = int(input("Enter index to delete: "))
if 0 <= idx < len(lst):
    del lst[idx]
    print("After deleting index:", lst)
else:
    print("Invalid index!")

if lst:
    lst.pop()
    print("After pop:", lst)
else:
    print("List is empty!")
```

```python
# Python Program to Illustrate Different Set Operations with User Input

set1 = set(map(int, input("Enter elements of Set1 (space-separated): ").split()))
set2 = set(map(int, input("Enter elements of Set2 (space-separated): ").split()))

print("Set1:", set1)
print("Set2:", set2)

print("Union:", set1 | set2)
print("Intersection:", set1 & set2)
print("Difference (Set1 - Set2):", set1 - set2)
print("Difference (Set2 - Set1):", set2 - set1)
print("Symmetric Difference:", set1 ^ set2)
print("Is Set1 subset of Set2?", set1.issubset(set2))
print("Is Set1 superset of Set2?", set1.issuperset(set2))
```

```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print("Select operation:")
print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")

choice = input("Enter choice (1/2/3/4): ")

if choice == '1':
    print("Result =", num1 + num2)
elif choice == '2':
    print("Result =", num1 - num2)
elif choice == '3':
    print("Result =", num1 * num2)
elif choice == '4':
    if num2 != 0:
        print("Result =", num1 / num2)
    else:
        print("Error: Division by zero")
else:
    print("Invalid choice")
```

```python
import string

text = input("Enter a string: ")
result = ""

for ch in text:
    if ch not in string.punctuation:
        result += ch

print("String without punctuation:", result)
```
